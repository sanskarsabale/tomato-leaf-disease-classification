```python
import os
import pandas as pd

RAW_DIR = "/content/raw_data/tomato"     # change if the classes are one level deeper

counts = {}
for cls in sorted(os.listdir(RAW_DIR)):
    cls_path = os.path.join(RAW_DIR, cls)
    if os.path.isdir(cls_path):
        counts[cls] = len(os.listdir(cls_path))

df = pd.DataFrame(list(counts.items()), columns=["class", "images"])
print(df.to_string(index=False))
print("\nNumber of classes:", len(df))
print("Total images:", df["images"].sum())
```

```python
import os
import pandas as pd

BASE = "/content/raw_data/tomato"
rows = []
for split in ["train", "val"]:
    for cls in sorted(os.listdir(f"{BASE}/{split}")):
        p = f"{BASE}/{split}/{cls}"
        if os.path.isdir(p):
            rows.append((split, cls, len(os.listdir(p))))

df = pd.DataFrame(rows, columns=["split", "class", "images"])
print(df.pivot(index="class", columns="split", values="images"))
print("\nTotals:\n", df.groupby("split")["images"].sum())
```

```python
from PIL import Image

bad = []
for split in ["train", "val"]:
    for cls in os.listdir(f"{BASE}/{split}"):
        d = f"{BASE}/{split}/{cls}"
        if not os.path.isdir(d):
            continue
        for f in os.listdir(d):
            try:
                with Image.open(f"{d}/{f}") as im:
                    im.verify()
            except Exception:
                bad.append(f"{d}/{f}")
print("Corrupted images:", len(bad))

# Leakage check: same file name in both train and val?
leaks = 0
for cls in os.listdir(f"{BASE}/train"):
    if os.path.isdir(f"{BASE}/val/{cls}"):
        leaks += len(set(os.listdir(f"{BASE}/train/{cls}")) & set(os.listdir(f"{BASE}/val/{cls}")))
print("File names shared between train and val:", leaks)
```

```python
import os, shutil, random
random.seed(42)

BASE = "/content/raw_data/tomato"

for cls in sorted(os.listdir(f"{BASE}/val")):
    src = f"{BASE}/val/{cls}"
    if not os.path.isdir(src):
        continue
    files = sorted(os.listdir(src))
    random.shuffle(files)
    half = len(files) // 2
    dst = f"{BASE}/test/{cls}"
    os.makedirs(dst, exist_ok=True)
    for f in files[:half]:
        shutil.move(f"{src}/{f}", f"{dst}/{f}")

for split in ["train", "val", "test"]:
    n = sum(len(os.listdir(f"{BASE}/{split}/{c}")) for c in os.listdir(f"{BASE}/{split}"))
    print(split, n)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications.resnet50 import preprocess_input

BASE = "/content/raw_data/tomato"
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42
AUTOTUNE = tf.data.AUTOTUNE

train_raw = keras.utils.image_dataset_from_directory(
    f"{BASE}/train", image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode="categorical", shuffle=True, seed=SEED)

val_raw = keras.utils.image_dataset_from_directory(
    f"{BASE}/val", image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode="categorical", shuffle=False)

test_raw = keras.utils.image_dataset_from_directory(
    f"{BASE}/test", image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode="categorical", shuffle=False)

class_names = train_raw.class_names
NUM_CLASSES = len(class_names)
print("Classes:", class_names)
print("Number of classes:", NUM_CLASSES)

# Augmentation for TRAINING images only
augment = keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])

# Train: augment, then ResNet50 preprocessing
train_ds = (train_raw
            .map(lambda x, y: (preprocess_input(augment(x, training=True)), y),
                 num_parallel_calls=AUTOTUNE)
            .prefetch(AUTOTUNE))

# Val and test: preprocessing only, no augmentation
val_ds = val_raw.map(lambda x, y: (preprocess_input(x), y),
                     num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)
test_ds = test_raw.map(lambda x, y: (preprocess_input(x), y),
                       num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)
```

```python
from tensorflow.keras.applications import ResNet50

# Pretrained ResNet50 without its original 1000-class top layer
base_model = ResNet50(weights="imagenet", include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False          # freeze the base for now

inputs = keras.Input(shape=(224, 224, 3))
x = base_model(inputs, training=False)        # training=False keeps BatchNorm layers in inference mode
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)

model = keras.Model(inputs, outputs)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)

model.summary()
```

```python
import os
os.makedirs("/content/models", exist_ok=True)

callbacks = [
    keras.callbacks.ModelCheckpoint(
        "/content/models/best_head.keras",
        monitor="val_accuracy", save_best_only=True, verbose=1),
    keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=4, restore_best_weights=True, verbose=1),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.3, patience=2, min_lr=1e-6, verbose=1),
]

history_head = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks,
)
```

```python
# Unfreeze only the last 30 layers of ResNet50
base_model.trainable = True
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep BatchNormalization layers frozen (safer for small fine-tuning)
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

# MUST recompile after changing trainable, with a much smaller learning rate
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)

trainable_count = sum(1 for l in model.layers[1].layers if l.trainable)
print("Trainable layers inside ResNet50:", trainable_count)

ft_callbacks = [
    keras.callbacks.ModelCheckpoint(
        "/content/models/best_finetuned.keras",
        monitor="val_loss", mode="min", save_best_only=True, verbose=1),
    keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=4, restore_best_weights=True, verbose=1),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.3, patience=2, min_lr=1e-7, verbose=1),
]

history_ft = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=ft_callbacks,
)
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

test_loss, test_acc = model.evaluate(test_ds, verbose=0)
print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_acc*100:.2f}%")

y_true = np.concatenate([np.argmax(y, axis=1) for _, y in test_ds])
y_prob = model.predict(test_ds, verbose=0)
y_pred = np.argmax(y_prob, axis=1)

short_names = [c.replace("Tomato___", "") for c in class_names]
print(classification_report(y_true, y_pred, target_names=short_names, digits=3))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=short_names, yticklabels=short_names)
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Confusion Matrix (Test Set)")
plt.xticks(rotation=45, ha="right"); plt.tight_layout()
plt.show()
```

```python
acc   = history_head.history["accuracy"]     + history_ft.history["accuracy"]
vacc  = history_head.history["val_accuracy"] + history_ft.history["val_accuracy"]
loss  = history_head.history["loss"]         + history_ft.history["loss"]
vloss = history_head.history["val_loss"]     + history_ft.history["val_loss"]
n_head = len(history_head.history["accuracy"])

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(acc, label="Train"); ax[0].plot(vacc, label="Validation")
ax[0].axvline(n_head - 1, color="gray", linestyle="--", label="Fine-tuning starts")
ax[0].set_title("Accuracy"); ax[0].set_xlabel("Epoch"); ax[0].legend()

ax[1].plot(loss, label="Train"); ax[1].plot(vloss, label="Validation")
ax[1].axvline(n_head - 1, color="gray", linestyle="--", label="Fine-tuning starts")
ax[1].set_title("Loss"); ax[1].set_xlabel("Epoch"); ax[1].legend()
plt.tight_layout(); plt.show()
```

```python
from google.colab import files
from tensorflow.keras.preprocessing import image as kimage
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.applications.resnet50 import preprocess_input

CONF_THRESHOLD = 0.70     # below this, we warn the user

def predict_leaf(path, top_k=3):
    img = kimage.load_img(path, target_size=(224, 224))
    arr = kimage.img_to_array(img)
    arr = preprocess_input(np.expand_dims(arr, axis=0))   # same preprocessing as training
    probs = model.predict(arr, verbose=0)[0]
    top_idx = np.argsort(probs)[::-1][:top_k]

    plt.figure(figsize=(4, 4))
    plt.imshow(img); plt.axis("off")
    best = class_names[top_idx[0]].replace("Tomato___", "")
    plt.title(f"{best}\n{probs[top_idx[0]]*100:.1f}%")
    plt.show()

    for i in top_idx:
        print(f"{class_names[i].replace('Tomato___', ''):45s} {probs[i]*100:5.1f}%")

    if probs[top_idx[0]] < CONF_THRESHOLD:
        print("\nWARNING: low confidence. This may not be a clear tomato leaf, "
              "or may be a disease the model was not trained on.")

uploaded = files.upload()
for name in uploaded:
    predict_leaf(name)
```
